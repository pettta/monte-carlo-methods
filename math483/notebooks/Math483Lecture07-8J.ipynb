```julia
using Distributions, Optim, Integrals, Random
```

```julia
#Lecture 7.1
#We use variance minimization to calculate the importance sampling shift
g(x) = (1-(x-5)^2)*(4<=x<=6) #Function we want the expected value of
f(x,M)=1/sqrt(2*pi)exp((x-M)^2/2) #N(M,1) pdf
g_VM(x,M)=(g(x))^2*(f(x,0))^2/f(x,M) #Integrand for the VM optimization problem
domain = (4,6) #Domain for the integral
p(M) = IntegralProblem(g_VM,domain,M)#These two lines of code define the integral function
I(M)=solve(p(M),QuadGKJL()).u  
optimize(I,1,10) #Now we search for a minimum of the parameter value M
```

```julia
#Lecture 7.2
#We use the relative entropy method to calculate the importance sampling shift
g(x) = (1-(x-5)^2)*(4<=x<=6) #Function we want the expected value of
f(x,M)=1/sqrt(2*pi)exp((x-M)^2/2) #N(M,1) pdf
g_RE(x,M)=-g(x)*f(x,0)log(f(x,M)) #Integrand for the VM optimization problem.  We put a - out front to change the maximization problem to minimization.
domain = (4,6) #Domain for the integral
p(M) = IntegralProblem(g_RE,domain,M)#These two lines of code define the integral function
I(M)=solve(p(M),QuadGKJL()).u  
optimize(I,1,10) #Now we search for a minimum of the parameter value M
```

```julia
#Lecture 7.3
#Now we test whether the results of the above optimization problems give better results than our intuition based IS sampler.
g(x) = (1-(x-5)^2)*(4<=x<=6) #The second parentheses return 0 unless x is in the desired range.
dist = Normal(0,1)
sample_size = 10000

#Importance Sampling with M=5 (intuition based shift)
g_IS(x,M) = (1-(x-5)^2)*exp(-x^2/2)*exp((x-M)^2/2)*(4<=x<=6)#IS distribution
IS_dist = Normal(5,1) #Importance sampling distribution
Y=rand(IS_dist,sample_size) 
I_IS=1/sample_size*sum(g_IS.(Y,5)) #Importance Sampling Estimator
V_IS=1/sample_size*sum((g_IS.(Y,5)-ones(sample_size)*I_IS).^2)/(sample_size-1)
println("Importance Sampling: ",I_IS)
println("Variance: ",V_IS)

#Importance Sampling with M=4.3 (approximate shift computed from VM/RE methods)
IS_dist_2 = Normal(4.3,1) #Importance sampling distribution
Y=rand(IS_dist_2,sample_size) 
I_IS_2=1/sample_size*sum(g_IS.(Y,4.3)) #Importance Sampling Estimator
V_IS_2=1/sample_size*sum((g_IS.(Y,4.3)-ones(sample_size)*I_IS).^2)/(sample_size-1)
println("Importance Sampling VM: ",I_IS_2)
println("Variance: ",V_IS_2)

println("True Expected Value: .0000112")
```
