```julia
using Random, Distributions
#Lecture 8.1
#We use a control variable to reduce the variance of a Monte Carlo estimator for the integral of sin(x^2) on [0,1].
#Crude Monte Carlo
g(x)=sin(x^2)
dist = Uniform(0,1)
sample_size = 10000
X=rand(dist,sample_size)
g_values = g.(X) #We'll use these later in the control variable computations
I_MC=1/sample_size*sum(g_values) #Monte Carlo Estimator
V_g = sum((g.(X)-ones(sample_size)*I_MC).^2)/(sample_size-1) #Variance of g(X)
V_MC=1/sample_size*V_g #Variance of I_MC
println("Crude_Monte Carlo: ",I_MC)
println("Variance: ",V_MC)

#Control Variable
C(x) = x^2 #First term in the Maclaurin series for sin(x)
C_values = C.(X)
E_C = 1/3 #Expected value of C(X)
V_C = 4/45 #Variance of C(X)
Cov = 1/(sample_size-1)*sum((g_values-ones(sample_size)*I_MC).*(C_values-E_C*ones(sample_size))) #Covariance between I_MC and C
k = Cov/V_C #Estimation of the optimal constant in the control variable estimator
println("Optimal k: ",k)
I_C = 1/sample_size*sum((g_values-k*(C_values-ones(sample_size)*E_C))) #Control Variable Estimator
V_est = 1/sample_size*sum((g_values-k*(C_values-ones(sample_size)*E_C)-ones(sample_size)*I_C).^2)/(sample_size-1)
V_thr = V_MC*(1-Cov^2/(V_C*V_g))
println("Control Variable Estimator: ",I_C)
println("Estimated Variance: ",V_est)


println("Actual Value of Integral: ",.310268)
```
