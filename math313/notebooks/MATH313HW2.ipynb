# Writing some brief R code for testing / getting the intervals from values given which scenario 

```r
# I am going to write three R functions corresponding to what we cover in the lecture so that 
# Once I write out what scenario applies to the problem, its easier to calculate 
# the confidence interval for the difference of means 

## Here's case 1 and 2 we covered in lecture (independent, normal, w and wout shared pop variance) 
conf_int_diff_means_same_var <- function(alpha, xbar, ybar, sx, sy, n, m) {
  df <- n+m-2 
  sp <- sqrt( (((n-1)*sx^2) + ((m-1)*sy^2) ) / (df))
  t_crit <- qt( alpha / 2, df, lower.tail=F)
  moe <- t_crit * sp * sqrt( (1/n) + (1/m ))
  lower <- (xbar - ybar) - moe
  upper <- (xbar - ybar) + moe
  return(c(lower=lower, upper=upper))
}

conf_int_diff_means_diff_var <- function(alpha, xbar, ybar, sx, sy, n, m) {
  df = trunc( ((sx^2 / n) + (sy^2 / m) )^2 / ( ((sx^2 / n)^2 / (n-1)) + ((sy^2 / m)^2 / (m-1))  ) ) 
  t_crit <- qt(alpha / 2, df, lower.tail=F) 
  moe <- t_crit * sqrt( (sx^2 / n)  + (sy^2 / m) )
  lower <- (xbar - ybar) - moe
  upper <- (xbar - ybar) + moe 
  return(c(lower=lower, upper=upper)) 
}

## Here's case 3 we covered in lecture (dependent normal ) 
conf_int_diff_means_dependent <- function(alpha, dbar, sd, n) {
  t_crit <- qt(alpha/2, n-1, lower.tail=F)
  moe <- t_crit * (sd / sqrt(n))
  lower <- dbar - moe 
  upper <- dbar + moe 
  return(c(lower=lower, upper=upper))
}
```

```r
# Verify that they work with the lecture problems 
conf_int_diff_means_same_var(.05, 10.26, 9.02, sqrt(6.32), sqrt(3.60), 10, 10)
conf_int_diff_means_diff_var(.05, 10.26, 9.02, sqrt(6.32), sqrt(3.60), 10, 10)
conf_int_diff_means_dependent(.05, 0.1987, 0.2383, 15) 
```
