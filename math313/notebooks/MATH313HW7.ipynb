# Question 6


```r
# Part a 
n <- c(15, 155, 110, 90, 35, 100, 30, 40, 75, 105)
e <- c(70, 45, 95, 95, 70, 315, 140, 260, 230, 400)
control <- c(535, 370, 420, 315, 485, 230, 370, 320, 335, 475)
group <- factor(c(rep('Norepinephrine', 10), rep('Epinephrine', 10), rep('Control', 10)))
measurement <- c(n, e, control)
data <- data.frame(group, measurement)
data
bartlett.test(measurement ~ group, data = data)
```

- since the p value = .02322 < alpha = .05, we reject h0. There is significant evidence at the alpha = .05 level that the variances are not equal. That means we should use the nonparametic analog to ANOVA which is Kruskal-Wallis 

```r
# Part b
kruskal.test(measurement ~ group, data = data)
```

- since the p value = .0001 < alpha = .05, we reject h0. There is significant evidence at the alpha = .05 level that at least one group differs from the others
