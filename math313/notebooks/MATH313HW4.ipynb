# Question 5


## Part a 
- $H_0: \mu \leq 1.0$
- $H_a: \mu > 1.0$

```r
# Part b 
data <- read.csv("fish_example.csv", header = T)
data
```

```r
# Part c
mu0 <- 1.0
n   <- nrow(data)
xbar <- mean(data$Mercury)
s   <- sd(data$Mercury)
df  <- n-1

# Perform a one‐sided t-test 
tt <- t.test(data$Mercury,
             mu = mu0,
             alternative = "greater",
             conf.level = 0.95)

tt
```

- p value is less than the value of alpha, so we reject the null hypothesis based on our findings 

# Question 6


- alternative hypothesis is $H_a: \mu \not = 170$

```r
students <- read.csv("student.csv", header=T)
students 
```

```r
mu0       <- 170      # H0: mean reaction time = 170 ms, Ha: not equal 
alpha     <- 0.05 
tt_react <- t.test(students$Reaction,
                   mu          = mu0,
                   alternative = "two.sided",
                   conf.level  = 1 - alpha)
tt_react
```

- Since we have a p-value of .00248 and alpha of .05, we reject the null hypothesis
- so we can conclude that the the true mean reaction differs from 170ms
